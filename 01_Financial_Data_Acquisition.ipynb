```python
import yfinance as yf
import pandas as pd
import requests
from bs4 import BeautifulSoup
from datetime import datetime, timedelta
```

## 1. Fetching Market Data
We use `yfinance` to grab historical price data. Let's look at CIPLA (a major Indian pharmaceutical stock, or any other ticker).


```python
ticker = 'CIPLA.NS' # Indian stock example
stock = yf.Ticker(ticker)

# Get last 1 month of 1-hour data
hist_data = stock.history(period='1mo', interval='1h')
hist_data.head()
```

## 2. Scraping Financial News
LLMs excel at unstructured text. Let's write a simple scraper to get recent news headlines from Yahoo Finance.


```python
def get_yahoo_finance_news(ticker):
    url = f'https://finance.yahoo.com/quote/{ticker}/news'
    headers = {'User-Agent': 'Mozilla/5.0'}
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # This is a very basic extraction; Yahoo's DOM changes often.
    # For production, consider using a dedicated News API like NewsData.io or AlphaVantage.
    headlines = []
    for h3 in soup.find_all('h3', class_='clamp'):
        headlines.append(h3.text.strip())
    return headlines

news = get_yahoo_finance_news('AAPL')
print("Recent Apple News:")
for n in news[:5]:
    print("-", n)
```
